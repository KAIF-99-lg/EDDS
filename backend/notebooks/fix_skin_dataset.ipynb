```
# Check what got downloaded in /tmp/skin
import os

for root, dirs, files in os.walk('/tmp/skin'):
    level = root.replace('/tmp/skin', '').count(os.sep)
    indent = ' ' * 2 * level
    print(f'{indent}{os.path.basename(root)}/')
    if level < 3:
        for f in files[:5]:
            print(f'{indent}  {f}')
        if len(files) > 5:
            print(f'{indent}  ... ({len(files)} total)')
```
